# Model evaluation
Runs the trained model against the held-out test split via `src/evaluation/evaluate.py` and displays the resulting figures inline. Requires `models/plastic_detector/best_model.pt` to exist (run `src/training/train.py` first).

```python
import sys, pathlib, json
sys.path.append(str(pathlib.Path.cwd().parent))
from src.evaluation import evaluate
import argparse

args = argparse.Namespace(
    weights='../models/plastic_detector/best_model.pt', data='../data/processed/data.yaml',
    split='test', imgsz=640, conf=0.25, device='auto', n_error_examples=12,
)
from ultralytics import YOLO
model = YOLO(args.weights)
summary = evaluate.run_official_metrics(model, args)
print(json.dumps(summary, indent=2))
```

```python
evaluate.sample_predictions(model, args)
evaluate.error_analysis(model, args)
```
